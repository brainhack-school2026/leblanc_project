```python
from pathlib import Path
import matplotlib.pyplot as plt
from PIL import Image
import ipywidgets as widgets
from IPython.display import display
```

```python
viz_path = Path.home() / "leblanc_project" / "results" / "brainmaps"

images = list(viz_path.glob("*.png"))

index = {}

for img in images:
    name = img.stem  # enlève .png
    pheno, atlas, comp, ses, p = name.split("_")

    index[(pheno, atlas, comp, ses, p)] = img
```

```python
phenotype = sorted(set(k[0] for k in index.keys()))
atlas = sorted(set(k[1] for k in index.keys()))
comparisons = sorted(set(k[2] for k in index.keys()))
sessions = sorted(set(k[3] for k in index.keys()))
pvalues = sorted(set(k[4] for k in index.keys()))
```

```python
def show_image(phenotype, atlas, comparison, session, pvalue):
    key = (phenotype, atlas, comparison, session, pvalue)

    if key in index:
        display(Image.open(index[key]))
    else:
        print("Image non trouvée")
```

```python
widgets.interact(
    show_image,
    phenotype=widgets.Dropdown(options=phenotype, description="Phenotype"),
    atlas=widgets.Dropdown(options=atlas, description="Atlas"),
    comparison=widgets.Dropdown(options=comparisons, description="Comparison"),
    session=widgets.Dropdown(options=sessions, description="Session"),
    pvalue=widgets.Dropdown(options=pvalues, description="P-value")
)
```
